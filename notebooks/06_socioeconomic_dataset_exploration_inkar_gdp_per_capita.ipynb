**Purpose:**

Selected as a socio-economic exposure indicator for the AI-Powered Climate & ESG Risk Intelligence Platform because economically important regions often contain greater concentrations of economic activity, infrastructure, businesses, employment, and financial assets that may be exposed to climate-related hazards. Combined with climate hazard indicators, GDP per capita contributes to assessing regional climate risk and supporting evidence-based decision making.

**Indicator type:** Socio-economic indicator (Economic Exposure)

**Spatial focus:** Germany (Districts / Kreise / NUTS-3)

**Dataset coverage:** 1995–2022

**Project temporal focus:** 1990–2025

### Temporal Harmonization by Boundary-Year Imputation

The Climate & ESG Risk Intelligence project adopts a common temporal reference period of 1990–2025 to ensure temporal consistency across all climate hazard and socioeconomic indicators included in the project.

The harmonized INKAR GDP per capita dataset provides annual observations for 1995–2022, resulting in a five-year gap at the beginning of the project period (1990–1994) and a three-year gap at the end (2023–2025).

To create a temporally consistent dataset for integration into the master Climate & ESG Risk Intelligence database, boundary-year imputation is applied using a conservative carry-backward and carry-forward approach. This method preserves the observed characteristics of the original INKAR dataset while avoiding the introduction of artificial economic trends or unsupported assumptions.

Specifically:

- 1990–1994 are populated using the earliest available observations from 1995.
- 2023–2025 are populated using the latest available observations from 2022.

As shown in the temporal trend analysis above, GDP per capita across German districts exhibits a clear long-term upward trend, reflecting sustained economic growth over the observation period, with only short-term fluctuations associated with broader economic conditions. However, reliable estimation of GDP per capita outside the observed period would require additional macroeconomic modelling, inflation adjustments, or economic forecasting assumptions. Rather than extrapolating the observed trend beyond the available data, this study adopts a conservative boundary-value imputation strategy in which the earliest available observations (1995) are used to populate 1990–1994 and the latest available observations (2022) are used to populate 2023–2025. This approach preserves the integrity of the original INKAR dataset, avoids introducing unsupported economic projections, and maintains methodological consistency across all socioeconomic indicators included in the Climate & ESG Risk Intelligence framework.

This harmonization strategy is applied consistently across all socioeconomic indicators to ensure methodological consistency, reproducibility, and compatibility with the project's unified temporal framework.

The temporally harmonized dataset generated in the following section forms the socioeconomic GDP per capita component of the integrated Climate & ESG Risk Intelligence master dataset.

```python
# =============================================================================
# Temporal Harmonization: Extend GDP per Capita Dataset to 1990–2025
# =============================================================================

print("=" * 80)
print("TEMPORAL HARMONIZATION: GDP PER CAPITA DATASET")
print("=" * 80)

# Check original temporal coverage
original_start_year = gdp_long["year"].min()
original_end_year = gdp_long["year"].max()

print(f"Original temporal coverage : {original_start_year}–{original_end_year}")
print(f"Original districts         : {gdp_long['district_code'].nunique()}")
print(f"Original observations      : {len(gdp_long):,}")

# -----------------------------------------------------------------------------
# 1. Backward Boundary-Year Imputation (1990–1994 using 1995 observations)
# -----------------------------------------------------------------------------

base_1995 = gdp_long[
    gdp_long["year"] == 1995
].copy()

backfill = []

for yr in range(1990, 1995):
    temp = base_1995.copy()
    temp["year"] = yr
    temp["imputation_flag"] = "backfilled_from_1995"
    backfill.append(temp)

backfill = pd.concat(backfill, ignore_index=True)

# -----------------------------------------------------------------------------
# 2. Forward Boundary-Year Imputation (2023–2025 using 2022 observations)
# -----------------------------------------------------------------------------

base_2022 = gdp_long[
    gdp_long["year"] == 2022
].copy()

forwardfill = []

for yr in [2023, 2024, 2025]:
    temp = base_2022.copy()
    temp["year"] = yr
    temp["imputation_flag"] = "forwardfilled_from_2022"
    forwardfill.append(temp)

forwardfill = pd.concat(forwardfill, ignore_index=True)

# -----------------------------------------------------------------------------
# 3. Add flag to observed INKAR data
# -----------------------------------------------------------------------------

gdp_observed = gdp_long.copy()
gdp_observed["imputation_flag"] = "observed_inkar"

# -----------------------------------------------------------------------------
# 4. Combine observed and imputed records
# -----------------------------------------------------------------------------

gdp_harmonized = pd.concat(
    [backfill, gdp_observed, forwardfill],
    ignore_index=True
)

gdp_harmonized = gdp_harmonized.sort_values(
    ["district_code", "year"]
).reset_index(drop=True)

# -----------------------------------------------------------------------------
# 5. Validate harmonized dataset
# -----------------------------------------------------------------------------

expected_years = list(range(1990, 2026))
expected_rows = 400 * len(expected_years)

print("\n" + "=" * 80)
print("VALIDATION AFTER TEMPORAL HARMONIZATION")
print("=" * 80)

print(f"Final temporal coverage : {gdp_harmonized['year'].min()}–{gdp_harmonized['year'].max()}")
print(f"Final districts         : {gdp_harmonized['district_code'].nunique()}")
print(f"Final observations      : {len(gdp_harmonized):,}")
print(f"Expected observations   : {expected_rows:,}")

print("\nRecords by imputation status:")
print(gdp_harmonized["imputation_flag"].value_counts())

assert gdp_harmonized["year"].min() == 1990
assert gdp_harmonized["year"].max() == 2025
assert gdp_harmonized["district_code"].nunique() == 400
assert len(gdp_harmonized) == expected_rows

duplicates = gdp_harmonized.duplicated(
    subset=["district_code", "year"]
).sum()

print(f"\nDuplicate district-year records: {duplicates}")
assert duplicates == 0

print("\n✓ Temporal harmonization completed successfully.")
```
