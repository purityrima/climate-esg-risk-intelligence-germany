**Description:**

This notebook uses the **INKAR (Indikatoren und Karten zur Raum- und Stadtentwicklung)** dataset **"Bruttoinlandsprodukt je Einwohner"**, which provides annual Gross Domestic Product (GDP) per capita statistics for approximately **400 German districts (Kreise)** covering the period **1995–2022**. GDP per capita is one of the most widely used indicators of regional economic performance and prosperity and serves as an important proxy for economic exposure, economic resilience, and adaptive capacity in the context of climate-related hazards.

**Purpose:**

Selected as a socio-economic exposure indicator for the AI-Powered Climate & ESG Risk Intelligence Platform because economically important regions often contain greater concentrations of economic activity, infrastructure, businesses, employment, and financial assets that may be exposed to climate-related hazards. Combined with climate hazard indicators, GDP per capita contributes to assessing regional climate risk and supporting evidence-based decision making.

**Indicator type:** Socio-economic indicator (Economic Exposure)

**Spatial focus:** Germany (Districts / Kreise / NUTS-3)

**Dataset coverage:** 1995–2022

**Project temporal focus:** 1990–2025

## Objective

To explore, evaluate, harmonize and preprocess the INKAR GDP per capita dataset for German districts (Kreise) in preparation for integration into the AI-Powered Climate & ESG Risk Intelligence Platform.

This notebook documents dataset characteristics, quality assessment, preprocessing decisions, visualization, temporal harmonization, and preparation for integration into the master Climate & ESG Risk Intelligence database.

### Import Libraries

```python
# ---------------------------------------------------------
# Climate & ESG Risk Intelligence Germany
# ---------------------------------------------------------
# Dataset Exploration
# Dataset: Real GDP per Capita (tipsna40)
# Indicator: Real GDP per Capita
# Category: Socio-economic Indicator (Economic Exposure)
# Author: Purity Mbaabu
# ---------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

### Locate Dataset

```python
# =========================================================
# Locate GDP per Capita dataset
# =========================================================

PROJECT_ROOT = Path.cwd().parent

gdp_file = (
    PROJECT_ROOT
    / "data"
    / "incoming"
    / "socioeconomic"
    / "GDP_per_Capita_INKAR_1995-2022-Regions400.xlsx"
)

print(gdp_file)
print(gdp_file.exists())
```

### Load Dataset

```python
# =========================================================
# Load GDP per Capita data
# =========================================================

gdp = pd.read_excel(
    gdp_file,
    sheet_name="Daten"
)

print(gdp.shape)
gdp.head()
```

##### Inspect workbook

```python
# =========================================================
# Inspect workbook
# =========================================================

xls = pd.ExcelFile(gdp_file)

print("Sheets:")
print(xls.sheet_names)
```

##### Preview the first few rows without headers

```python
# =========================================================
# Preview raw workbook structure
# =========================================================

preview = pd.read_excel(
    gdp_file,
    sheet_name="Daten",
    header=None,
    nrows=8
)

preview
```

##### Inspect dataframe structure

```python
# =========================================================
# Inspect dataframe structure
# =========================================================

print(f"Rows: {gdp.shape[0]}")
print(f"Columns: {gdp.shape[1]}")

print("\nColumn names:")
print(gdp.columns.tolist())
```

##### Clean the column names

```python
# =========================================================
# Harmonize column names
# =========================================================

new_columns = [
    "district_code",
    "district_name",
    "aggregation_level"
]

# Years stored in first data row
years = gdp.iloc[0, 3:].astype(int).astype(str).tolist()

new_columns.extend(years)

gdp.columns = new_columns

gdp.head()
```

##### Remove the year row

```python
# =========================================================
# Remove year row
# =========================================================

gdp = gdp.iloc[1:].reset_index(drop=True)

print(gdp.shape)

gdp.head()
```

##### Verify district count

```python
# =========================================================
# Verify district coverage
# =========================================================

print("Aggregation levels:")
print(gdp["aggregation_level"].value_counts())

print()

print("Number of districts:")
print(len(gdp))
```

##### Check for duplicate district codes

```python
# ==========================================================
# Check duplicate district codes
# ==========================================================

duplicates = gdp["district_code"].duplicated().sum()

print("Duplicate district codes:", duplicates)
```

##### Check missing values

```python
# ==========================================================
# Missing values
# ==========================================================

missing = gdp.isna().sum()

print(missing[missing > 0])
```

##### Convert district codes to official 5-digit strings

```python
# ==========================================================
# Format district codes
# ==========================================================

gdp["district_code"] = (
    gdp["district_code"]
        .astype(int)
        .astype(str)
        .str.zfill(5)
)

gdp.head()
```

##### Check the conversion

```python
print(gdp["district_code"].head())
```

##### Format district codes


Convert district identifiers to the official five-digit German AGS (Amtlicher Gemeindeschlüssel) format to ensure compatibility with climate, administrative boundary, and socio-economic datasets.

```python
# ==========================================================
# Format district codes
# ==========================================================

gdp["district_code"] = (
    gdp["district_code"]
        .astype(int)
        .astype(str)
        .str.zfill(5)
)

gdp.head()
```

##### Verify district codes

```python
# ==========================================================
# Verify district codes
# ==========================================================

print(gdp["district_code"].head())

print()
print("Unique district codes:", gdp["district_code"].nunique())
```

##### Convert from wide to long format


Transform annual GDP columns into a single `year` column and a single `gdp_per_capita` column.

This tidy format is required for merging with climate indicators and for machine learning workflows.

```python
# ==========================================================
# Convert wide table to long format
# ==========================================================

gdp_long = gdp.melt(
    id_vars=[
        "district_code",
        "district_name",
        "aggregation_level"
    ],
    var_name="year",
    value_name="gdp_per_capita"
)

gdp_long.head()
```

##### Convert data types

```python
# ==========================================================
# Convert data types
# ==========================================================

gdp_long["year"] = gdp_long["year"].astype(int)

gdp_long["gdp_per_capita"] = (
    pd.to_numeric(gdp_long["gdp_per_capita"], errors="coerce")
)
```

##### Verify the transformed dataset

```python

print(gdp_long.shape)

gdp_long.head()
```

##### Quality checks

```python
# ==========================================================
# Quality checks after transformation
# ==========================================================

print("Missing values")
print(gdp_long.isna().sum())

print()

print("Duplicate rows")
print(gdp_long.duplicated().sum())

print()

print("Unique districts")
print(gdp_long["district_code"].nunique())

print()

print("Years")
print(
    gdp_long["year"].min(),
    "-",
    gdp_long["year"].max()
)
```

##### Save the cleaned dataset

```python
# ==========================================================
# Save cleaned GDP dataset
# ==========================================================

output_file = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "socioeconomic"
    / "gdp_per_capita_inkar_1995_2022.csv"
)

gdp_long.to_csv(output_file, index=False)

print(output_file)
```
