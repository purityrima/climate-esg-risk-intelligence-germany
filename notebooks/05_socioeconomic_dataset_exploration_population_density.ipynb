## Objective

To explore, evaluate, harmonize, and preprocess the Population Density dataset for Germany in preparation for integration into the AI-Powered Climate & ESG Risk Intelligence Platform.

This notebook documents dataset characteristics, quality assessment, preprocessing decisions, visualization, temporal harmonization, and preparation for integration into the master Climate & ESG Risk Intelligence database.

### Import Libraries

```python
# ============================================================
# Climate & ESG Risk Intelligence Germany
# Dataset Exploration 05
# Dataset: Population Density (demo_r_d3dens)
# Indicator: Population Density
# Category: Socio-economic Indicator (Exposure)
# Author: Purity Mbaabu
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
```

### Locate Dataset

```python
# ============================================================
# Locate Population Density dataset
# ============================================================

from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

population_file = (
    PROJECT_ROOT
    / "data"
    / "incoming"
    / "socioeconomic"
    / "demo_r_d3dens__custom_22023791_spreadsheet.xlsx"
)

print(population_file)
print(population_file.exists())
```

### Load Dataset

```python
# ============================================================
# Load Population Density data from Sheet 1
# ============================================================

population = pd.read_excel(
    population_file,
    sheet_name="Sheet 1",
    header=7
)

population.head()
```

##### Clean Eurostat Excel structure

```python
# ============================================================
# Clean Eurostat Excel structure
# ============================================================

# Rename first two columns
population = population.rename(
    columns={
        "TIME": "GEO (Codes)",
        "TIME.1": "GEO (Labels)"
    }
)

# Remove Eurostat duplicated header row
population = population[
    population["GEO (Codes)"] != "GEO (Codes)"
].copy()

# Clean GEO code column
population["GEO (Codes)"] = population["GEO (Codes)"].astype(str).str.strip()

# Check result
print(population.shape)
population.head()
```

#### Step 2. Filter German only

```python
# ============================================================
# Filter Germany observations
# ============================================================

population_de = population[
    population["GEO (Codes)"].str.startswith("DE", na=False)
].copy()

print("Germany-only shape:", population_de.shape)

population_de.head()
```

##### Identify the NUTS levels.

```python
# ============================================================
# Identify NUTS levels based on GEO code length
# ============================================================

population_de["nuts_level"] = population_de["GEO (Codes)"].str.len()

population_de["nuts_level"].value_counts().sort_index()
```

**Comment**
- Climate datasets: 401 districts
- Population dataset: 407 districts

This difference is expected and worth documenting. It is usually caused by changes in Germany's administrative boundaries over the 1990–2024 period. Eurostat preserves historical NUTS units, whereas the climate datasets are harmonized to a single NUTS version.

##### Keep only NUTS-3

```python
# ============================================================
# Keep German NUTS-3 districts only
# ============================================================

population_nuts3 = population_de[
    population_de["nuts_level"] == 5
].copy()

print("German NUTS-3 shape:", population_nuts3.shape)

population_nuts3[["GEO (Codes)", "GEO (Labels)"]].head(20)
```

##### Inspect Eurostat quality flags

Eurostat exports one additional flag column after every annual value column. These columns contain observation quality annotations (e.g., e, b, ep) rather than population density values. They are inspected and documented before being removed from the analytical dataset.

```python
# ============================================================
# Inspect Eurostat quality flag columns
# ============================================================

flag_columns = [col for col in population_nuts3.columns
                if "Unnamed" in str(col)]

print("Number of quality flag columns:", len(flag_columns))
print(flag_columns)
```

##### Remove Eurostat quality flag columns

The quality flag columns are not numerical observations and are therefore removed. The remaining dataset contains only the geographical identifiers and annual population density values.

```python
# ============================================================
# Remove Eurostat quality flag columns
# ============================================================

population_nuts3 = population_nuts3.drop(columns=flag_columns)

print(population_nuts3.shape)

population_nuts3.head()
```

##### Inspect the available years

```python
# ============================================================
# Identify available year columns
# ============================================================

year_cols = [
    col for col in population_nuts3.columns
    if str(col).isdigit()
]

print(year_cols)

print("First year :", year_cols[0])
print("Last year  :", year_cols[-1])
print("Total years:", len(year_cols))
```

##### Replace Eurostat flags with missing values

**Replace Eurostat quality flags**

Eurostat uses special symbols such as ":" and alphabetic flags to describe
missing or provisional observations. These are converted to standard
missing values (NaN) before assessing temporal completeness.

```python
# ============================================================
# Replace Eurostat flags with NaN
# ============================================================

import numpy as np

population_nuts3[year_cols] = (
    population_nuts3[year_cols]
    .replace(
        {
            ":": np.nan,
            "e": np.nan,
            "b": np.nan,
            "be": np.nan,
            "ep": np.nan,
            "bep": np.nan
        }
    )
)

population_nuts3.head()
```

```python
# ============================================================
# Number of available observations per year
# ============================================================

population_nuts3[year_cols].notna().sum()
```

**Comment**

Observation
The Germany NUTS-3 population density dataset does not provide complete temporal coverage across all districts. No district-level observations are available between 1990 and 1996. Between 1997 and approximately 2013, the number of districts with reported values gradually increases from 372 to 401, indicating evolving data completeness and/or historical administrative boundary revisions reflected in the Eurostat dataset. From 2014 onwards, the dataset is nearly complete (400–401 districts). These temporal gaps and inconsistencies will be addressed during the preprocessing stage through temporal harmonization and imputation.

##### Identify which districts are missing

```python
# ============================================================
# Inspect missing districts at selected years
# ============================================================

for year in ["1997", "2000", "2010", "2024"]:

    missing = population_nuts3[
        population_nuts3[year].isna()
    ][["GEO (Codes)", "GEO (Labels)"]]

    print("=" * 60)
    print(f"{year}: {len(missing)} missing districts")
    display(missing.head(20))
```

##### Count observations per district

Identify the NUTS-3 districts that have complete population density observations throughout the available period (1997–2024).

```python
# ============================================================
# Number of available years per district
# ============================================================

population_nuts3["available_years"] = population_nuts3[year_cols].notna().sum(axis=1)

population_nuts3["available_years"].describe()
```

##### Inspect the distribution

```python
population_nuts3["available_years"].value_counts().sort_index()
```

##### Extract the stable districts

```python
# ============================================================
# Districts with complete observations (1997-2024)
# ============================================================

stable_population = population_nuts3[
    population_nuts3["available_years"] == 28
].copy()

print("Stable districts:", stable_population.shape[0])

stable_population[["GEO (Codes)", "GEO (Labels)"]].head(20)
```

##### Verify

```python
stable_population["1997"].notna().sum()
```

```python
stable_population["2024"].notna().sum()
```

##### Save stable reference districts

```python
stable_population_codes = stable_population["GEO (Codes)"].tolist()

len(stable_population_codes)

# save the data

stable_population.to_csv(
    "../data/raw/socioeconomic/stable_population_districts_1997_2024.csv",
    index=False
)
```

**Comment**

Finding

Of the 407 German NUTS-3 districts contained in the Eurostat population density dataset, 364 districts have complete observations for every year from 1997–2024. The remaining districts exhibit temporal gaps primarily due to administrative boundary changes and the evolution of the NUTS classification over time.

Rather than imputing values for districts that were created, merged, or redefined through administrative reforms, this project retains the 364 stable districts as the reference population panel. These districts will later be intersected with the stable districts identified in the GDP and unemployment datasets to construct a balanced socio-economic panel for subsequent integration with the climate datasets.
